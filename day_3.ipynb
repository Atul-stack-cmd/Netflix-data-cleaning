### Encapsulation Demo
In this example, we'll create a `BankAccount` class where the account balance `__balance` is private so it cannot be modified directly from outside. Instead, we use methods like `deposit()`, `withdraw()`, and a getter method `get_balance()` to interact with it safely.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner          # Public attribute
        self.__balance = balance    # Private attribute (starts with double underscore)

    # Getter method to read the private attribute safely
    def get_balance(self):
        return self.__balance

    # Method to update the private attribute with validation
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: ${amount}")
        else:
            print("Invalid deposit amount!")

    # Method to withdraw with validation
    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew: ${amount}")
        else:
            print("Insufficient funds or invalid amount!")
```

```python
# Create an instance
acct = BankAccount("Alice", 1000)

# Access public attribute directly
print("Owner:", acct.owner)

# Try to access private attribute directly (This will raise an AttributeError)
try:
    print(acct.__balance)
except AttributeError as e:
    print("Error:", e)

# Use proper getter to view the balance
print("Initial Balance:", acct.get_balance())

# Modify via approved methods
acct.deposit(500)
acct.withdraw(200)
print("Final Balance:", acct.get_balance())
```

### Operator Overloading Demo
Let's create a class `Point` that represents 2D coordinates `(x, y)`. We will overload the addition operator (`+`) and the string representation (`__str__`) so we can easily add points together and print them.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overload the + operator
    def __add__(self, other):
        new_x = self.x + other.x
        new_y = self.y + other.y
        return Point(new_x, new_y)

    # Overload string representation for friendly printing
    def __str__(self):
        return f"Point({self.x}, {self.y})"
```

```python
p1 = Point(1, 2)
p2 = Point(3, 4)

# Using the + operator on custom objects!
p3 = p1 + p2

print("p1:", p1)
print("p2:", p2)
print("Result of p1 + p2:", p3)
```

### Abstraction Demo
Abstraction allows us to define a common blueprint (an abstract class) for a set of subclasses. An abstract class cannot be instantiated directly; it contains one or more **abstract methods** that any concrete subclass must implement.

```python
from abc import ABC, abstractmethod

# Abstract Class (Blueprint)
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

    @abstractmethod
    def stop_engine(self):
        pass

# Concrete Subclass 1
class Car(Vehicle):
    def start_engine(self):
        print("Car: Engine started. Vroom!")

    def stop_engine(self):
        print("Car: Engine stopped.")

# Concrete Subclass 2
class Motorcycle(Vehicle):
    def start_engine(self):
        print("Motorcycle: Engine started. Braap!")

    def stop_engine(self):
        print("Motorcycle: Engine stopped.")
```

```python
# Try to instantiate the abstract class (This will raise a TypeError)
try:
    v = Vehicle()
except TypeError as e:
    print("Error:", e)

# Create instances of concrete classes
my_car = Car()
my_bike = Motorcycle()

my_car.start_engine()
my_bike.start_engine()
```
